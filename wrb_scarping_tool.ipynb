```python
# Importation des bibliothèques nécessaires
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv("data/data_ws/coin_afric.csv")
```

```python
df.head()
```

```python
df['adresse'].unique()
```

```python
df = df.drop_duplicates().reset_index(drop=True)

# Suppression des devise et conversion en float
df["prix"] = df["prix"].str.replace("CFA", "", regex=False)
df["prix"] = df["prix"].str.replace(" ", "", regex=False)
df["prix"] = pd.to_numeric(df["prix"], errors="coerce")

# Renplacement des valeurs manquantes du ['prix'] par la moyenne
if df["prix"].isna().any():
    df["prix"] = df["prix"].fillna(df["prix"].mean())

# Remplacement des valeur manquante de ['adresse'] par le mode
if df["adresse"].isna().any():
    df["adresse"] = df["adresse"].fillna(df["adresse"].mode()[0])

#Suppression des 2 première collonne
df = df.drop(columns=["web_scraper_order", "web_scraper_start_url"], errors="ignore")



```

```python
df.rename(columns={df.columns[0]: "categorie"}, inplace=True)
```
